```python
%load_ext autoreload
%autoreload 2

from catrace.dataset import load_dataset_config
dsconfig = load_dataset_config('../dataset_configs/juvenile_dataset_t7.json')
```

```python
import pandas as pd
from catrace.geometry import participation_ratio
def compute_all_manifold_participation_ratios(dff):
    pr = participation_ratio(dff.values)
    return pr

def compute_each_manifold_participation_ratios(dff, odors, manifold_level='manifold'):
    prs = []
    for odor in odors:
        pr = participation_ratio(dff.xs(odor, level=manifold_level).values)
        prs.append(pr)

    pr_df = pd.DataFrame(prs, index=odors)
    return pr_df

def compute_manifold_centers_participation_ratios(dff, manifold_level='manifold'):
    centers = dff.groupby(level=manifold_level).mean()
    pr_centers = participation_ratio(centers)
    return pr_centers

from catrace.process_time_trace import select_time_points

def compute_exp_participation_ratios(dff, window, odors, manifold_level='manifold'):
    dff_window = select_time_points(dff, window)
    pr_all = compute_all_manifold_participation_ratios(dff_window)
    pr_each = compute_each_manifold_participation_ratios(dff_window, odors, manifold_level=manifold_level)
    pr_centers = compute_manifold_centers_participation_ratios(dff_window, manifold_level=manifold_level)
    pr_summary = pr_each.T
    pr_summary['all'] = pr_all
    pr_summary['centers'] = pr_centers
    return pr_summary
```

```python
from catrace.exp_collection import read_df
trace_dir = dsconfig.processed_trace_dir
exp_list = dsconfig.exp_list

window = [38, 76]

pr_summaries = []
for exp_name, _ in exp_list:
    dff = read_df(trace_dir, exp_name)
    pr_summary = compute_exp_participation_ratios(dff, window, dsconfig.odors_stimuli, manifold_level='odor')
    pr_summaries.append(pr_summary)
```

```python

all_pr_df = pd.concat(pr_summaries, keys=exp_list, names=['fish_id', 'condition'])
```

```python
import os
pr_dir = os.path.join(dsconfig.processed_trace_dir, 'participation_ratio')
os.makedirs(pr_dir, exist_ok=True)
all_pr_df.to_pickle(os.path.join(pr_dir, f'participation_ratio_all_fish_window{window[0]}_{window[1]}.pkl'))
```

```python
import matplotlib.pyplot as plt
from catrace.visualize import plot_measure_by_cond, PlotBoxplotByCondParams

color_dict = {'naive': 'tab:blue', 
              'arg-phe': 'salmon',
              'phe-arg': 'm',
              'phe-trp': 'tab:green',
}
plot_params = PlotBoxplotByCondParams(
    figsize=(2.0, 1.8),
    box_linewidth=1,
    mean_marker_size=0.7,
    strip_size=0.5,
    ylim=None,#(0, 0.18),
    pvalue_marker_xoffset=0.2,
    box_colors= color_dict.values(),
    ylevel_scale=0.95,
)
fig, ax = plt.subplots(figsize=(1.6,1.5))
plot_measure_by_cond(all_pr_df_sorted, measure_name='centers', params=plot_params, ax=ax)

tick_interval = 2
ax.yaxis.set_major_locator(plt.MultipleLocator(tick_interval))
ax.set_ylabel('Participation ratio \nof the space spanned \nby manifold centers')
ax.set_ylim((0, 14))

dataset_name = 'juvenile'
window_name = 'time_window_5s'
fig_dir = f'../../../publish_code/figures/{dataset_name}/{window_name}/participation_ratio'

import os
os.makedirs(fig_dir, exist_ok=True)
fig.tight_layout()
fig.savefig(os.path.join(fig_dir, 'participation_ratio_space_spanned_by_manifold_centers.pdf'))
```

```python
import seaborn as sns
figsize=(2.5, 1.8)
fig, ax = plt.subplots(figsize=figsize)
datadf = within_manifold_pr_df.reset_index(level=['fish_id', 'condition'])
hue_colors = list(color_dict.values())
strip_hue_colors = ['gray', 'gray', 'gray', 'gray']
mean_hue_colors = hue_colors
do_plot_strip = True
condition_name = 'condition'
odor_name = 'odor_category'
yname = 'participation_ratio'


test_results=None
ylabel=None
ylim=None
label_fontsize=7
legend_fontsize=7
hline_y=None
box_width=0.5
box_linewidth=1
do_plot_strip=True
strip_size=1
strip_jitter=0.2
strip_alpha=0.9
box_hue_separation_scaler=1.5
strip_hue_separation_scaler=0.95
mean_dodge=0.55
mean_marker_size=2
pvalue_marker_fontsize=7
pvalue_marker_xoffset=0.01
pvalue_bar_linewidth=1
do_capitalize_labels=False
show_ns=False


if ax is None:
    fig, ax = plt.subplots(
        figsize=figsize
    )  # Adjusted size for better visibility
else:
    fig = ax.get_figure()
nconds = datadf[condition_name].nunique()

if mean_hue_colors is None:
    mean_hue_colors = [_get_darker_color(color) for color in hue_colors]
# Plotting stripplot and boxplot with hue
if do_plot_strip:
    sns.stripplot(
        ax=ax,
        x=odor_name,
        y=yname,
        hue=condition_name,
        data=datadf,
        jitter=strip_jitter,
        dodge=True,
        size=strip_size,
        alpha=strip_alpha,
        zorder=1,
        palette=strip_hue_colors,
        hue_separation_scaler=strip_hue_separation_scaler,
    )
sns.boxplot(
    ax=ax,
    x=odor_name,
    y=yname,
    hue=condition_name,
    data=datadf,
    saturation=0.5,
    width=box_width,
    zorder=2,
    dodge=True,
    medianprops=dict(alpha=0.95, linewidth=box_linewidth),
    boxprops=dict(alpha=0.95, linewidth=box_linewidth),
    whiskerprops=dict(linewidth=box_linewidth, alpha=0.7),
    showfliers=False,
    showcaps=False,
    fill=False,
    palette=hue_colors,
    hue_separation_scaler=box_hue_separation_scaler,
)

# Add mean points
mean_points = datadf.groupby(
    [odor_name, condition_name], as_index=False, sort=False
)[yname].mean()
sns.pointplot(
    ax=ax,
    x=odor_name,
    y=yname,
    hue=condition_name,
    data=mean_points,
    dodge=mean_dodge,
    markers='D',
    linestyle='none',
    zorder=3,
    markersize=mean_marker_size,
    palette=mean_hue_colors,
)


handles, labels = ax.get_legend_handles_labels()
# Keep only the last N entries of the legend, N = len(dsconfig.conditions)
handles = handles[-nconds:]
labels = labels[-nconds:]
ax.legend(handles, labels, fontsize=legend_fontsize)

ax.set_ylabel('Participation ratio \nof odor manifold', fontsize=7)
ax.set_xlabel('Odor category', fontsize=7)
sns.despine(ax=ax)
ax.set_ylim((0,14))
tick_interval = 2
ax.yaxis.set_major_locator(plt.MultipleLocator(tick_interval))
# Set tick fontsize=7 for x and y
ax.tick_params(axis='both', which='major', labelsize=7)

fig.tight_layout()
fig.savefig(os.path.join(fig_dir, 'participation_ratio_within_odor_manifold.pdf'))
```
