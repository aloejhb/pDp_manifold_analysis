```python
# Normalize and fit with exponential
import pandas as pd
import numpy as np
from scipy.optimize import curve_fit

# Assuming your DataFrame is named df and columns are time points
# Example: df = pd.DataFrame({...})
df = ccdf
# Step 1: Select time points from 5200 to 8000
time_points = df.columns.astype(int)
selected_time_points = (time_points >= 5000) & (time_points <= 8000)
df_selected = df.loc[:, selected_time_points]

# Step 2: Normalize the current values by the average between 5000 and 5200
baseline_time_points = (time_points >= 5000) & (time_points <= 5200)
baseline_means = df.loc[:, baseline_time_points].mean(axis=1)
df_normalized = df_selected.div(baseline_means, axis=0)

# Step 3: Define the exponential decay function
def exponential_decay(t, A, tau, C):
    return A * np.exp(-t / tau) + C

# Step 4: Perform exponential fit for each neuron and extract the time constant (tau)
time_constants = []

for index, row in df_normalized.iterrows():
    selected_time_points_without_first200 = (time_points >= 5200) & (time_points <= 8000)
    t = time_points[selected_time_points_without_first200] - 5200  # Shift time to start from 0
    y = row.values[200:]
    
    # Initial guess for the parameters [A, tau, C]
    p0 = [1, 1000, 0]
    
    try:
        # Fit the exponential decay function
        popt, _ = curve_fit(exponential_decay, t, y, p0=p0)
        tau = popt[1]  # Extract the time constant (tau)
        time_constants.append(tau)
    except RuntimeError:
        # If the fit fails, append NaN
        time_constants.append(np.nan)

# Step 5: Summarize the time constants in a DataFrame
time_constants_df = pd.DataFrame(time_constants, columns=['Time Constant'])

# Step 6: Calculate and print the mean, median, and std
mean_tau = time_constants_df['Time Constant'].mean()
median_tau = time_constants_df['Time Constant'].median()
std_tau = time_constants_df['Time Constant'].std()

print(f"Mean Time Constant: {mean_tau}")
print(f"Median Time Constant: {median_tau}")
print(f"Standard Deviation of Time Constants: {std_tau}")
```
