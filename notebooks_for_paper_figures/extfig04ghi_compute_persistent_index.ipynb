```python
from catrace.process_time_trace import select_odors_df
from catrace.process_time_trace import select_time_points

def compute_delta_index(dff, odor, offset_boundary, pre_offset_window_size, post_offset_window_size, frame_rate, cut_time):
    """
    Compute the delta index for a given odor.
    
    Parameters:
    - dff: DataFrame containing the dF/F traces for all neurons and odors.
    - odor: The odor for which to compute the persistence index (e.g., 'odor1' or 'odor2').
    - offset_boundary: Time (in seconds) that defines the boundary between pre-offset and post-offset periods.
    - pre_offset_window_size: Size of the window (in seconds) to consider for the pre-offset period.
    - post_offset_window_size: Size of the window (in seconds) to consider for the post-offset period.
    - frame_rate: Frame rate of the recordings (in Hz).
    - cut_time: Time (in seconds) that was cut from the beginning of the traces.
    
    Returns:
    - delta_index: A Series containing the delta index for each neuron.
    """
    # Calculate frame indices for pre-offset and post-offset periods
    offset_boundary_frame = int((offset_boundary + cut_time) * frame_rate)
    pre_offset_start_frame = offset_boundary_frame - int(pre_offset_window_size * frame_rate)
    pre_offset_end_frame = offset_boundary_frame
    post_offset_start_frame = offset_boundary_frame
    post_offset_end_frame = offset_boundary_frame + int(post_offset_window_size * frame_rate)
    
    # Ensure that frame indices are within bounds
    pre_offset_start_frame = max(pre_offset_start_frame, 0)
    post_offset_end_frame = min(post_offset_end_frame, dff.shape[0])
    
    # Select dF/F traces for the specified odor
    odor_dff = select_odors_df(dff, [odor])
    
    # Compute mean dF/F in pre-offset and post-offset windows
    # pre_offset_mean = select_time_points(odor_dff, range(pre_offset_start_frame, pre_offset_end_frame)).mean()
    # post_offset_mean = select_time_points(odor_dff, range(post_offset_start_frame, post_offset_end_frame)).mean()
    pre_offset_median = select_time_points(odor_dff, range(pre_offset_start_frame, pre_offset_end_frame)).median()
    post_offset_median = select_time_points(odor_dff, range(post_offset_start_frame, post_offset_end_frame)).median()

    # Compute delta index
    delta_index = (post_offset_median - pre_offset_median) / (pre_offset_median + post_offset_median + 1e-6)  # Add small value to avoid division by zero
    # Absolute value of delta index to capture both increases and decreases in activity
    delta_index = delta_index.abs()
    
    return delta_index, pre_offset_median, post_offset_median
```

```python
import pandas as pd
offset_boundary = 7
pre_offset_window_size = 2
post_offset_window_size = 1
#pre_mean_threshold = 0.5
#pre_mean_threshold = 1.5
pre_median_threshold = 1.5

odors = dsconfig.odors_stimuli

# Compute delta index for each odor and store in a dataframe
delta_indices = {}
for odor in odors:
    delta_ind, pre_offset_median, post_offset_median = compute_delta_index(dff, odor, offset_boundary=offset_boundary, pre_offset_window_size=pre_offset_window_size, post_offset_window_size=post_offset_window_size, frame_rate=dsconfig.frame_rate, cut_time=cut_time)

    # Set the rows where pre_offset_median is below the threshold to NaN in the delta index
    delta_ind[pre_offset_median < pre_median_threshold] = float('nan')
    delta_indices[odor] = delta_ind


delta_indices_df = pd.DataFrame(delta_indices)

persistent_indices_df = 1 - delta_indices_df

# If the entire row is nan, remove the row
persistent_indices_df = persistent_indices_df.dropna(how='all')
# Fill rest of the nan's with zero
persistent_indices_df = persistent_indices_df.fillna(0)
```

```python
def cluster_neurons(index_df, n_clusters=4, random_state=42):
    # Scale
    from sklearn.preprocessing import StandardScaler
    scaler = StandardScaler()
    index_df_scaled = scaler.fit_transform(index_df.T).T
    from sklearn.cluster import KMeans
    # Perform K-means clustering
    kmeans = KMeans(n_clusters=n_clusters, random_state=random_state)
    cluster = kmeans.fit_predict(index_df_scaled)
    clustered_df = index_df.copy()
    clustered_df['cluster'] = cluster
    clustered_scaled_df = pd.DataFrame(index_df_scaled, index=index_df.index, columns=index_df.columns)
    clustered_scaled_df['cluster'] = cluster
    return clustered_df, clustered_scaled_df
```

```python
# Test whether naive and trained are different in normalized_count in yes_pers_normalized_count
# trained includes all conditions except naive
from scipy.stats import mannwhitneyu
naive_counts = yes_pers_normalized_count_df[yes_pers_normalized_count_df['condition'] == 'naive']['normalized_count']
trained_counts = yes_pers_normalized_count_df[yes_pers_normalized_count_df['condition'] != 'naive']['normalized_count']
stat, pvalue = mannwhitneyu(naive_counts, trained_counts, alternative='two-sided')
print(f'Mann-Whitney U test between naive and trained for normalized count of neurons with 1 persistent odor: U={stat}, p={pvalue:.4f}')
```
