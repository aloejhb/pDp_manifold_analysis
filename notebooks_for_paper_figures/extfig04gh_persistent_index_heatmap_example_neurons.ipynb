```python
odor_colors = ['#d25dd2', '#766acb', '#428cba', '#da866b', '#daa36b', '#879a73']
pallete = dict(zip(odors, odor_colors))
import seaborn as sns
def plot_example_neuron_trace(neuron_id, dff, dsconfig, cut_time, offset_boundary, title=None, ax=None):
    if ax is None:
        fig, ax = plt.subplots(figsize=(1.4, 0.6))
    else:
        fig = ax.get_figure()
    # Find the column index of the neuron in dff
    col = dff.columns.get_loc(neuron_id)

    # Retrieve the neuron trace
    neuron_trace = dff.iloc[:, col]
    neuron_trace_df = neuron_trace.reset_index()

    # Use seaborn to plot the neuron trace with time on x-axis and dF/F on y-axis, and color the line by odor
    #fig, ax = plt.subplots(figsize=(6, 3))
    # average over trials and only keep time and odor
    neuron_trace_df = neuron_trace_df.groupby(['time', 'odor']).mean()
    # Rename the 'time' column to 'frame'
    neuron_trace_df = neuron_trace_df.reset_index().rename(columns={'time': 'frame'})
    # Add a 'time' column by dividing the 'frame' column by the frame rate
    neuron_trace_df['time'] = neuron_trace_df['frame'] / dsconfig.frame_rate - cut_time
    sns.lineplot(data=neuron_trace_df, x='time', y=neuron_trace.name, hue='odor', ax=ax, linewidth=0.8, palette=pallete)
    # ax.set_ylabel('Estimated firing rate (Hz)', fontsize=7)
    # ax.set_xlabel('Time (s)', fontsize=7)
    # Remove xlabel and ylabel
    ax.set_xlabel('')
    ax.set_ylabel('')
    # Remove xticks
    #ax.set_xticks([])
    # if title is not None:
    #     ax.set_title(title, fontsize=7)
    # else:
    #     ax.set_title(f'Neuron {neuron_id} Trace', fontsize=7)
    # set tick fondsize
    ax.tick_params(axis='both', which='major', labelsize=7)
    # Put a vertical line at the offset boundary
    ax.axvline(x=offset_boundary, color='gray', linestyle='--', label='Offset Boundary', linewidth=1)
    # Remove legend
    ax.legend().remove()
    # Don't show x tick labels
    # ax.set_xticklabels([])
    # Set x and y tick length to 2
    ax.tick_params(axis='both', which='major', length=2)
    sns.despine()
    plt.tight_layout()
    return fig, ax, neuron_trace_df

examples_row_num = [6, 485, 1004, 1423, 1600, 1803, 2206]
#[200, 700, 1000, 1400, 1600, 1800, 2200]
# Create subplots with 1 column and len(examples_row_num) rows
fig, axes = plt.subplots(len(examples_row_num), 1, figsize=(1.4, 0.5 * len(examples_row_num)), sharex=True)
neuron_trace_dfs = []
for i, row_num in enumerate(examples_row_num):
    neuron_id = pat.iloc[row_num].name
    fig, ax, neuron_trace_df = plot_example_neuron_trace(neuron_id, dff, dsconfig, cut_time, offset_boundary, title=row_num, ax=axes[i])
    if i == 4:
        # Set y tick interval to 2
        from matplotlib.ticker import MultipleLocator
        ax.yaxis.set_major_locator(MultipleLocator(2))
    # save_figure_for_paper(fig, f'above_thresh_example_neuron_{row_num}', paper_fig_dir)
    neuron_trace_dfs.append(neuron_trace_df)
plt.tight_layout()
# save_figure_for_paper(fig, f'above_thresh_example_neuron_all', paper_fig_dir)
```

```python
neuron_trace_dfs[0].set_index(['time', 'frame', 'odor', 'trial'])

neuron_trace_set_index_dfs = [pd.DataFrame(df).set_index(['time', 'frame', 'odor', 'trial']) for df in neuron_trace_dfs]

# Concatenate the dataframes along the columns (axis=1)
concatenated_neuron_trace_df = pd.concat(neuron_trace_set_index_dfs, axis=1)
concatenated_neuron_trace_df

# Set names for column levels to fish_id, condition, plane, neuron, neuron_id
concatenated_neuron_trace_df.columns.names = ['fish_id', 'condition', 'plane', 'neuron', 'neuron_id']
concatenated_neuron_trace_df

# Save as CSV
concatenated_neuron_trace_df.to_csv(os.path.join(save_dir, 'extended_figure04h_example_neuron_traces.csv'))
```
